```python
# Define the exact hierarchical order
severity_mapping = {'minor': 0, 'major': 1, 'fatal': 2}
density_mapping = {'low': 0, 'medium': 1, 'high': 2}
visibility_mapping = {'low': 0, 'medium': 1, 'high': 2}

# Apply the mapping to overwrite the text with numbers
df['accident_severity'] = df['accident_severity'].map(severity_mapping)
df['traffic_density'] = df['traffic_density'].map(density_mapping)
df['visibility'] = df['visibility'].map(density_mapping)
```

### Step 5: Clean dates and convert categories
I changed the date column into a proper datetime format so I could pull out month and year. Then I converted category labels like 'minor', 'major', and 'fatal' into numbers so the model can understand them better.

```python
# List the columns that have no inherent order
unranked_columns = ['cause', 'weather', 'road_type', 'day_of_week']

# Convert them into binary columns (drop_first=True prevents multicollinearity)
df = pd.get_dummies(df, columns=unranked_columns, drop_first=True)
```

### Step 6: Turn text categories into numbers
Some columns like cause and weather are text values, so I converted them into separate binary columns using one-hot encoding. This helps machine learning models understand the data without mixing words together.
